### D

```python
cinemas_df.select(
    "cinema_name",
    F.col("location.zone").alias("zone")
).show()
```

```python
facilities_df = cinemas_df.select(
    "cinema_id",
    "cinema_name",
    F.explode("facilities").alias("facility")
)
facilities_df.show()
```

```python
dated_df = bookings_df.withColumn(
    "booking_date",
    F.to_date("booking_date")
).withColumn(
    "show_date",
    F.to_date("show_date")
).withColumn(
    "days_between_booking_and_show",
    F.datediff("show_date", "booking_date")
)
```

```python
dated_df.select(
    "booking_id",
    "booking_date",
    "show_date",
    "days_between_booking_and_show"
).show()
```

```python
duplicates_df = bookings_df.groupBy("booking_id").count().filter(
    F.col("count") > 1
)

duplicates_df.show()
```

```python
deduplicated_df = bookings_df.dropDuplicates(["booking_id"])
deduplicated_df.show()
```

### E

```python
movie_revenue_df = bookings_df.groupBy("movie_title").agg(
    F.sum(F.col("tickets")*F.col("price_per_ticket")).alias("net_revenue")
)
```

```python
from pyspark.sql import Window
revenue_window = Window.orderBy(F.col("net_revenue").desc())
```

```python
ranked_movies_df = movie_revenue_df.withColumn(
    "revenue_rank",
    F.rank().over(revenue_window)
)
ranked_movies_df.show()
```

```python
final_df = deduplicated_df.join(
    cinemas_df,
    on="cinema_id",
    how="left"
)
```

```python
final_df.write.format("delta").mode("overwrite").saveAsTable(
    "cinema_bookings_cleaned"
)
```

```python
display(
    spark.sql("""
        SELECT *
        FROM cinema_bookings_cleaned
        LIMIT 5
    """)
)
```

### Final Challenge

```python
report_base_df = dated_df.dropDuplicates(["booking_id"])
```

```python
report_base_df = report_base_df.filter(
    F.col("booking_status") == "Confirmed"
)
```

```python
report_joined_df = report_base_df.join(
    cinemas_df,
    on="cinema_id",
    how="left"
)
```

```python
performance_report_df = report_joined_df.groupBy(
    F.col("location.city").alias("city"),
    "movie_title"
).agg(
    F.sum("tickets").alias("total_tickets"),
    F.sum(F.col("tickets")*F.col("price_per_ticket")).alias("net_revenue"),
    F.count("booking_id").alias("total_bookings")
)
```

```python
performance_report_df = performance_report_df.orderBy(
    F.col("net_revenue").desc()
)
```

```python
performance_report_df.show()
```

```python
performance_report_df.write.format("delta").mode("overwrite").saveAsTable(
    "cinema_performance_report"
)
```
