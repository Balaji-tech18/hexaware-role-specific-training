```python
%sql
SELECT *
FROM fitness_memberships_demo;

```

```python
%sql
SELECT member_name, city, membership_plan
FROM fitness_memberships_demo;
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
WHERE city = 'Hyderabad';
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
WHERE monthly_fee > 2000;
```

```python
%sql
SELECT DISTINCT city
FROM fitness_memberships_demo;
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
WHERE membership_plan IN ('Premium', 'Standard');
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
WHERE monthly_fee BETWEEN 1000 AND 2000;
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
WHERE member_name LIKE 'S%';
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
WHERE status = 'Active'
  AND city = 'Hyderabad';
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
WHERE city = 'Pune'
   OR city = 'Chennai';
```

```python
%sql
SELECT
    member_name AS Name,
    monthly_fee AS Fee
FROM fitness_memberships_demo;
```

```python
%sql
SELECT m.*
FROM fitness_memberships_demo AS m
WHERE m.membership_plan = 'Premium';
```

### Part C

```python
%sql
DESCRIBE HISTORY fitness_memberships_demo;
```

```python
%sql
SELECT *
FROM fitness_memberships_demo VERSION AS OF 1;
```

```python
%sql
SELECT
    'Earlier Version' AS snapshot,
    *
FROM fitness_memberships_demo VERSION AS OF 1
UNION ALL
SELECT
    'Current Version' AS snapshot,
    *
FROM fitness_memberships_demo;
```

```python
%sql
SELECT COUNT(*) AS remaining_members
FROM fitness_memberships_demo;
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
ORDER BY member_id;
```
